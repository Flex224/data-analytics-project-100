```python
import pandas as pd
import requests 
```

```python
base_url = 'https://data-charts-api.hexlet.app'

params = {
    
    'begin': '2023-03-01',
    'end': '2023-09-01'
    
}
```

```python
visits_df = pd.read_csv('visits_1k.csv')
regs_df = pd.read_csv('regs_1k.csv')
```

```python
response_regs = requests.get(base_url + '/registrations', params=params)

print(f"Статус запроса регистраций: {response_regs.status_code}")
```

```python
data_visits = response_visits.json()
data_regs = response_regs.json()
```

```python
visits_df_api = pd.DataFrame(data_visits)
regs_df_api = pd.DataFrame(data_regs)
```

```python
print(visits_df_api.head())
```

```python
print(regs_df_api.head())
```

```python
visits_df_api['datetime'] = pd.to_datetime(visits_df_api['datetime'])
regs_df_api['datetime'] = pd.to_datetime(regs_df_api['datetime'])

print("Даты преобразованы.")
```

```python
visits_cleaned = visits_df_api[~visits_df_api['user_agent'].str.contains('bot', case=False)]
print(f"Было визитов: {len(visits_df_api)}, стало без ботов: {len(visits_cleaned)}")
```

```python
visits_cleaned = visits_cleaned.sort_values(by='datetime')
visits_cleaned = visits_cleaned.drop_duplicates(subset='visit_id', keep='last')

print(f"Осталось уникальных визитов: {len(visits_cleaned)}")
```

```python
visits_cleaned['date_group'] = visits_cleaned['datetime'].dt.date

# (Не забудь сделать 'date_group' и для регистраций)
regs_df_api['date_group'] = regs_df_api['datetime'].dt.date

print("\nКолонка 'date_group' создана в обоих DataFrame'ах.")

# Проверяем, что получилось
print("--- Визиты (проверка) ---")
print(visits_cleaned[['datetime', 'date_group']].head())

print("\n--- Регистрации (проверка) ---")
print(regs_df_api[['datetime', 'date_group']].head())
```

```python
visits_grouped = visits_cleaned.groupby(['date_group', 'platform'])['visit_id'].count().reset_index(name='visits')
```
